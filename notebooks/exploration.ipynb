```python
import pandas as pd
import matplotlib.pyplot as plt

motif_summary = pd.read_csv(
    "../results/motif_summary.csv"
)
motif_positions = pd.read_csv(
    "../results/motif_positions.csv"
)

motif_summary.head()
```

```python
print("Summary shape:", motif_summary.shape)
print("Positions shape:", motif_positions.shape)
```

```python
motif_totals = (
    motif_summary
    .groupby("motif_name")["count"]
    .sum()
    .sort_values(ascending=False)
)
print(motif_totals)
```

```python
motif_summary.groupby(
    "motif_name"
)["density_per_kb"].describe()
```

```python
family_density = (
    motif_summary
    .groupby(["class", "motif_name"])
    ["density_per_kb"]
    .mean()
    .unstack()
)
family_density
```
